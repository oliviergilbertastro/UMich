```python
from datasets import get_dataset_config_names, get_dataset_split_names, load_dataset

REPO = "nyu-planet-finders/kepler-raw-lightcurves"

print("Checking available configurations...")
configs = get_dataset_config_names(REPO)
print("Configurations:", configs)

for config in configs:
    print(f"\nChecking splits for {config!r}...")
    splits = get_dataset_split_names(REPO, config_name=config)
    print("Splits:", splits)

print("\nStreaming one light-curve row...")
lightcurves = load_dataset(
    REPO,
    "lightcurves",
    split="full",
    streaming=True,
)

row = next(iter(lightcurves))
print("Light-curve fields:", list(row.keys()))
print("Field types:", {key: type(value).__name__ for key, value in row.items()})

for key, value in row.items():
    if isinstance(value, list):
        print(f"{key}: {len(value)} values; first three = {value[:3]}")

print("\nStreaming one TCE catalog row...")
tces = load_dataset(
    REPO,
    "tces",
    split="full",
    streaming=True,
)

tce = next(iter(tces))
print("TCE fields:", list(tce.keys()))
print("TCE values:", tce)
```
