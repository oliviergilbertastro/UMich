Show the results in ascending order of the month

Hint: Some of the functions that can help are rank(), unstack(), stack(). 

```python
new_df = df[["carrier", "month", "dep_delay"]]

# Compute average departure delay for each carrier in each month
avg_delay = new_df.groupby(["month", "carrier"])["dep_delay"].mean()
#print(avg_delay["9E"])
# Rank carriers within each month (lowest delay = rank 1)
ranked = avg_delay.groupby(level="month").rank(method="min")

# Keep only the carriers ranked first
result = avg_delay[ranked == 1].reset_index()

# Sort by month
result = result.sort_values("month")
avg_delay = result
print(result)
```

## Question 1.8 (4 pts)
Segment the dep_time 24 hours into 
* night from 12 midnight to 5 am
* morning from 5:01 am to 12 noon
* afternoon from 12.01 noon to 5 pm
* evening from 5.01 pm to 12 am

And then run a query to find the total flights in each segment and save that value in the given variable.


```python
bins = [0, 500, 1200, 1700, 2400]
labels = ["night", "morning", "afternoon", "evening"]

df["part_of_day"] = pd.cut(
    df["dep_time"],
    bins=bins,
    labels=labels,
    right=True,
    include_lowest=True
)

parts_of_day_count = df["part_of_day"].value_counts()
print(parts_of_day_count['morning'])
print(parts_of_day_count['evening'])
```

```python
assert parts_of_day_count['morning'] == 129939
assert parts_of_day_count['evening'] == 98362
```

## Question 2.1 (3 pts)

Line charts are ideal to show a trend line. Draw a line chart to show monthly frequency count of flights as a trend line for the year 2013. Be sure to choose the correct data to display on the x-axis and y-axis.

```python
monthly_flights = flights.groupby("month").size().reset_index(name="flight_count")

fig1 = sns.lineplot(data=monthly_flights, x="month", y="flight_count", marker="o")

fig1.set_title("Monthly breakdown")
fig1.set_xlabel("Months (1-12)")
fig1.set_ylabel("Monthly Frequency")
```

## Question 2.3 (3 pts)

Segment the flights based on time-slots with the following criteria
* flights that depart from 00-5:59am as 'red-eye' flights
* flights that depart from 6am - 11:59am as 'morning'
* flights that depart from 12noon - 3:59pm as 'afternoon'
* flights that depart from 4pm - 11:59pm as 'evening' 

And using seaborn create a plot as shown below. PS: Feel free to choose your color palette.

segmentation.png

```python
bins = [0, 559, 1159, 1559, 2359]
labels = ["red-eye", "morning", "afternoon", "evening"]

flights["time_slot"] = pd.cut(
    flights["dep_time"],
    bins=bins,
    labels=labels,
    right=True,
    include_lowest=True
)

titles = [
    "red-eye(00-5:59am)",
    "morning(6am-11:59am)",
    "afternoon(12noon-3:59pm)",
    "evening(4pm-11:59pm)"
]

fig_3 = sns.catplot(
    data=flights,
    x="origin",
    col="time_slot",
    col_order=labels,
    kind="count",
    palette="viridis",
    height=4,
    aspect=0.8
)

for ax, title in zip(fig_3.figure.get_axes(), titles):
    ax.set_title(title)
    ax.set_ylabel("Flight Count")
    ax.set_xlabel("")

# It seems like the plot is only EWR, JFK, and LGA, but the instructions don't say to cut to these...

flights_short = flights[flights["origin"].isin(["EWR", "JFK", "LGA"])]

flights_short.describe()

titles = [
    "red-eye(00-5:59am)",
    "morning(6am-11:59am)",
    "afternoon(12noon-3:59pm)",
    "evening(4pm-11:59pm)"
]

fig_3 = sns.catplot(
    data=flights_short,
    x="origin",
    col="time_slot",
    col_order=labels,
    kind="count",
    palette="viridis",
    height=4,
    aspect=0.8
)

for ax, title in zip(fig_3.figure.get_axes(), titles):
    ax.set_title(title)
    ax.set_ylabel("Flight Count")
    ax.set_xlabel("")
```

```python
assert fig_3.figure.get_axes()[0].get_title() == 'red-eye(00-5:59am)'
assert fig_3.figure.get_axes()[0].get_ylabel() == 'Flight Count'
assert fig_3.figure.get_axes()[1].get_title() == 'morning(6am-11:59am)'
```
