```python
import numpy as np 
import matplotlib.pyplot as plt
import astropy.units as u
import astropy.constants as cst

L_star = 2E5*u.Lsun

T_sub_silicate = 1200*u.K
T_sub_graphite = 2000*u.K

def get_distance(T):
    return np.sqrt(L_star/(16*np.pi*cst.sigma_sb*T**4))

dist_silicate = get_distance(T_sub_silicate).to(u.AU)
dist_graphite = get_distance(T_sub_graphite).to(u.AU)

print(dist_silicate)
print(dist_graphite)

Q0 = 1E49*u.s**-1
alpha_BH = 3.1E-13*(u.cm**3/u.s)
n_H = 2000*u.cm**-3

R_stromgren = (3*Q0/(4*np.pi*n_H**2*alpha_BH))**(1/3)

print(R_stromgren.to(u.AU))
print(R_stromgren.to(u.lyr))
```
