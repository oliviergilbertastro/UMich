```python
import numpy as np 
import matplotlib.pyplot as plt
import astropy.units as u
import astropy.constants as cst

L_star = 2E5*u.Lsun

T_sub_silicate = 1200*u.K
T_sub_graphite = 2000*u.K

def get_distance(T):
    return np.sqrt(L_star/(16*np.pi*cst.sigma_sb*T**4))

dist_silicate = get_distance(T_sub_silicate).to(u.AU)
dist_graphite = get_distance(T_sub_graphite).to(u.AU)

print(dist_silicate)
print(dist_graphite)

Q0 = 1E49*u.s**-1
alpha_BH = 3.1E-13*(u.cm**3/u.s)
n_H = 2000*u.cm**-3

R_stromgren = (3*Q0/(4*np.pi*n_H**2*alpha_BH))**(1/3)

print(R_stromgren.to(u.AU))
print(R_stromgren.to(u.lyr))

rho = 1.225E-3*u.g*u.cm**-3
n_H = rho/cst.m_p
print(n_H.to(u.cm**-3))


ell = 1.086*1.9E21*u.cm**-2/n_H
print(ell.to(u.cm))
```

```python
import astropy.units as u

# Given parameters
n_H = 40 / u.cm**3
T = 80 * u.K
x = 1e-3

# Reaction rates
k_ra = 1.6e-16 * u.cm**3 / u.s * (T / (80 * u.K))**0.67
k_ad = 4e-9 * u.cm**3 / u.s
zeta_det = 2.7e-7 / u.s
zeta_dis = 4.2e-11 / u.s

# H- equilibrium abundance
f_Hminus = (
    x * n_H * k_ra
    / (zeta_det + n_H * (1 - x) * k_ad)
).decompose()

# H2 molecular fraction
f_mol = (
    2 * f_Hminus * n_H * (1 - x) * k_ad
    / zeta_dis
).decompose()

print(f"f_Hminus = {f_Hminus.value:.3e}")
print(f"f_mol = {f_mol.value:.3e}")
```
