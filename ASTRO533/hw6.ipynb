```python
import astropy.units as u

n_H = 40 * u.cm**-3
x = 1e-3

n_Hminus_over_n_H = (
    x * n_H * 1.6e-16 * u.cm**3 / u.s
    / (
        2.7e-7 / u.s
        + n_H * (1 - x) * 4e-9 * u.cm**3 / u.s
    )
)

print(n_Hminus_over_n_H)
```

```python
import astropy.units as u
import astropy.constants as cst
import numpy as np
T_A=25*u.K
x = np.log(1+1/(T_A/(5.532*u.K)+0.151))

T_exc = 5.532*u.K/x
print(T_exc)
```

```python

T_A=9.1*u.K
nu = 110.2*u.GHz
b = 3*u.km/u.s
tau_0 = 2.845E-21*u.cm**2*(1-np.exp(-cst.h*nu/(cst.k_B*T_exc)))*(cst.c/b)
tau_0 = tau_0.to(u.cm**2)
print(tau_0)

T_A_eq = cst.h*nu/cst.k_B* ( 1/(np.exp(cst.h*nu/(cst.k_B*T_exc))-1) - 1/(np.exp(cst.h*nu/(cst.k_B*2.725*u.K))-1) )*tau_0

print(T_A_eq.to(u.K*u.cm**2))

N_0 = T_A/T_A_eq

print(N_0.to(u.cm**-2))
```

```python
B_0 = cst.h*nu/2
Q_exc = np.sum([(2*J+1)*np.exp(-B_0*J*(J+1)/(cst.k_B*T_exc)) for J in range(100)])

print(Q_exc)
f_0 = 1/Q_exc
print(f_0)
print((N_0/f_0).to(u.cm**-2))

N_0_CO = (N_0/f_0).to(u.cm**-2)
```
